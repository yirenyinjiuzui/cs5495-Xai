````python



from pathlib import Path
import os, time, random, tarfile, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset
from torchvision import models, datasets, transforms
from torchvision.transforms import functional as TF
from PIL import Image, ImageEnhance, ImageFilter

# if use Google Drive
# from google.colab import drive
# drive.mount('/content/drive')

BASE_DIR = Path('/content') if Path('/content').is_dir() else Path('/mnt/data')
ARCHIVE_PATH = BASE_DIR / 'imagenette2-320.tgz'  #  Drive Path (if want)
WEIGHT_PATH  = BASE_DIR / 'resnet18-f37072fd.pth'
DATA_ROOT    = BASE_DIR / 'imagenette2-320'
OUTPUT_DIR   = BASE_DIR / 'xai_outputs'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
QUICK_TEST = False
BATCH_SIZE = 64 if torch.cuda.is_available() else 32
NUM_WORKERS = 2
SEED = 2026
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
if DEVICE.type == 'cpu':
    torch.set_num_threads(min(4, os.cpu_count() or 1))
print(f'Device={DEVICE}, PyTorch={torch.__version__}, output={OUTPUT_DIR}')
print(f'Weight found: {WEIGHT_PATH.is_file()}; archive found: {ARCHIVE_PATH.is_file()}')
assert WEIGHT_PATH.is_file(), f'Can not find the model weights, please check WEIGHT_PATH: {WEIGHT_PATH}'
assert ARCHIVE_PATH.is_file() or (DATA_ROOT/'val').is_dir(), f'Can not find the dataset, please check ARCHIVE_PATH: {ARCHIVE_PATH}'
````

## 1 解压数据与模型预处理
在实验中数据使用 `val`，评估泛化性能，使用ResNet18官方 ImageNet-1K V1预处理：短边 Resize 到 256 → CenterCrop 224 → ToTensor → ImageNet 均值/标准差归一化。类别 ID 映射是必要步骤：`ImageFolder` 的 10 类本地标签（0–9）不能与 ResNet18 1000 类输出直接对比。

````python
if not (DATA_ROOT/'val').is_dir():
    print('Extracting', ARCHIVE_PATH.name)
    with tarfile.open(ARCHIVE_PATH, 'r:gz') as tar:
        members = tar.getmembers()
        for m in members:
            p = Path(m.name)
            if p.is_absolute() or '..' in p.parts or not (m.isfile() or m.isdir()):
                raise ValueError(f'The zip file contains unsafe entries: {m.name}')
        try:
            tar.extractall(path=BASE_DIR, members=members, filter='data')
        except TypeError:
            tar.extractall(path=BASE_DIR, members=members)
print('Validation set directory:', DATA_ROOT/'val')

WEIGHTS_META = models.ResNet18_Weights.IMAGENET1K_V1
TRANSFORM = WEIGHTS_META.transforms()
IMAGENET_CLASSES = WEIGHTS_META.meta['categories']

# The order is the same as the WNID alphabetical order in ImageFolder, and the assertion prevents data labels from silently misaligning.
WNID_TO_NAME = {
    'n01440764': 'tench',
    'n02102040': 'English springer',
    'n02979186': 'cassette player',
    'n03000684': 'chain saw',
    'n03028079': 'church',
    'n03394916': 'French horn',
    'n03417042': 'garbage truck',
    'n03425413': 'gas pump',
    'n03445777': 'golf ball',
    'n03888257': 'parachute',}
VAL_DATASET = datasets.ImageFolder(str(DATA_ROOT/'val'), transform=TRANSFORM)
assert set(VAL_DATASET.classes) == set(WNID_TO_NAME), 'The directory categories do not match the official Imagenette categories'
CLASS_NAMES_10 = [WNID_TO_NAME[k] for k in VAL_DATASET.classes]
IMAGENET_ID_BY_LOCAL = [IMAGENET_CLASSES.index(n) for n in CLASS_NAMES_10]
assert IMAGENET_ID_BY_LOCAL == [0, 217, 482, 491, 497, 566, 569, 571, 574, 701]
assert len(VAL_DATASET) == 3925, f' output {len(VAL_DATASET)} '
print('Full validation set：', len(VAL_DATASET), '；ImageNet output 1000 class')
print(pd.DataFrame({'WNID': VAL_DATASET.classes, 'Imagenette label': range(10),
                    'ImageNet-1K label': IMAGENET_ID_BY_LOCAL, 'class': CLASS_NAMES_10}).to_string(index=False))

MODEL = models.resnet18(weights=None)
try:
    state = torch.load(WEIGHT_PATH, map_location='cpu', weights_only=True)
except TypeError:
    state = torch.load(WEIGHT_PATH, map_location='cpu')
MODEL.load_state_dict(state, strict=True)
MODEL = MODEL.to(DEVICE).eval()
assert MODEL.fc.out_features == 1000
print('Model loaded successfully, output dimensions：', MODEL.fc.out_features)
````

## 2完整验证集预测与模型准确率


````python
# 2.1 Predict 1000 classes
EVAL_DATASET = Subset(VAL_DATASET, range(min(256,len(VAL_DATASET)))) if QUICK_TEST else VAL_DATASET
EVAL_INDICES = list(range(len(EVAL_DATASET)))
loader = DataLoader(EVAL_DATASET, batch_size=BATCH_SIZE, shuffle=False,num_workers=NUM_WORKERS, pin_memory=(DEVICE.type=='cuda'))
rows = []; started = time.time()
with torch.inference_mode():
    current = 0
    for images, local_targets in loader:
        images = images.to(DEVICE, non_blocking=True)
        probabilities = MODEL(images).softmax(dim=1)
        scores5, ids5 = probabilities.topk(5, dim=1)
        for j, local_id in enumerate(local_targets.tolist()):
            idx = EVAL_INDICES[current+j]
            true_id = IMAGENET_ID_BY_LOCAL[local_id]
            pred_id = int(ids5[j,0].item())
            top5_ids = ids5[j].tolist()
            rows.append({
                'dataset_index': idx,
                'filename': str(Path(VAL_DATASET.samples[idx][0]).relative_to(DATA_ROOT)),
                'true_imagenette_id': local_id,
                'true_imagenet1k_id': true_id,
                'true_class': CLASS_NAMES_10[local_id],
                'pred_imagenet1k_id': pred_id,
                'pred_class': IMAGENET_CLASSES[pred_id],
                'pred_confidence': float(scores5[j,0].item()),
                'top5_imagenet1k_ids': ' '.join(map(str, top5_ids)),
                'top1_correct': int(pred_id == true_id),
                'top5_correct': int(true_id in top5_ids),})
        current += len(local_targets)
PREDICTIONS = pd.DataFrame(rows)
TOP1 = PREDICTIONS['top1_correct'].mean()
TOP5 = PREDICTIONS['top5_correct'].mean()
print(f'sample N={len(PREDICTIONS)}，Top-1={TOP1:.4%}，Top-5={TOP5:.4%}，time {time.time()-started:.1f}s')
if QUICK_TEST:
    print('Now is QUICK_TEST')
else:
    print(' Finish 3925 picture evaluation')
PREDICTIONS.to_csv(OUTPUT_DIR/'resnet18_imagenette_val_predictions.csv', index=False)
display(PREDICTIONS.head(8))
````

## 3 Feature Map Visualization and Class Activation Mapping (CAM)

### 3.1 原理与解释对象

本实验通过可视化 ResNet18 的中间特征图，观察不同通道对输入图像的空间响应；进一步采用 CAM，将卷积特征与分类器权重结合，构造针对预测类别的空间重要性热图，该实现属于基于已有输入的激活可视化与类别归因

````python
class FeatureVisualizationCAM:
    def __init__(self, model, layer=None):
        self.model = model
        self.layer = layer if layer is not None else model.layer4
        if not (hasattr(model, 'fc') and hasattr(model.fc, 'weight')):
            raise ValueError('CAM requires a GAP followed by a linear classification layer; the current implementation is for ResNet18')

    @torch.no_grad()
    def extract(self, input_tensor, target_class):
        if input_tensor.ndim != 4 or input_tensor.shape[0] != 1:
            raise ValueError('This interface only processes one image at a time: [1,3,H,W]')
        activations = []
        handle = self.layer.register_forward_hook(lambda m, i, o: activations.append(o.detach()))
        try:
            logits = self.model(input_tensor)
        finally:
            handle.remove()
        feats = activations[0][0]  # [512,7,7]
        weights = self.model.fc.weight[int(target_class)]
        if feats.shape[0] != weights.numel():
            raise ValueError('The number of CAM feature channels does not match the classifier weights')
        signed_cam = torch.einsum('c,chw->hw', weights, feats)
        cam = signed_cam.clamp_min(0)
        energy = feats.square().mean(dim=0).sqrt()

        def resize_normalize(two_d):
            up = F.interpolate(two_d[None,None], size=input_tensor.shape[-2:],
                               mode='bilinear', align_corners=False)[0,0]
            up = up - up.min()
            return (up / up.max().clamp_min(1e-8)).cpu().numpy().astype(np.float32)

        return {'features': feats.cpu().numpy(),
                'energy': resize_normalize(energy),
                'cam': resize_normalize(cam),
                'prediction': int(logits.argmax(dim=1).item())}

FV = FeatureVisualizationCAM(MODEL)
MEAN = torch.tensor([0.485,0.456,0.406]).view(1,3,1,1)
STD  = torch.tensor([0.229,0.224,0.225]).view(1,3,1,1)

def show_rgb(input_tensor):
    """Denormalization for visualization, without modifying the input tensor."""
    a = input_tensor.detach().cpu().float()
    rgb = (a * STD + MEAN).clamp(0,1)
    return rgb[0].permute(1,2,0).numpy()

def fv_explainer(input_tensor, target_class):
    """[Unified Interface] XAI(input_tensor[1,3,H,W], int target) -> Non-negative [H,W] importance map."""
    return FV.extract(input_tensor, target_class)['cam']

# Pick the first correctly classified image from each different category
VIS_INDICES = []
for local_class in range(10):
    found = PREDICTIONS[(PREDICTIONS.true_imagenette_id==local_class) &
                        (PREDICTIONS.top1_correct==1)]
    if len(found): VIS_INDICES.append(int(found.iloc[0].dataset_index))
VIS_INDICES = VIS_INDICES[:6]
assert VIS_INDICES, 'Without correctly predicted samples, it is impossible to create a visual demonstration.'
print('Feature Visualization sample index：', VIS_INDICES)
````

````python
# 3.3 Different Categories: Original Image / Feature Intensity / CAM / Overlay
fig, axes = plt.subplots(len(VIS_INDICES), 4, figsize=(13, 3.1*len(VIS_INDICES)), squeeze=False)
for row_idx, idx in enumerate(VIS_INDICES):
    x, local_target = VAL_DATASET[idx]
    x = x.unsqueeze(0).to(DEVICE)
    target = int(PREDICTIONS.loc[PREDICTIONS.dataset_index==idx, 'pred_imagenet1k_id'].iloc[0])
    result = FV.extract(x, target)
    rgb = show_rgb(x)
    for col, image in enumerate([rgb, result['energy'], result['cam'], rgb]):
        axis = axes[row_idx,col]
        if col in [0,3]: axis.imshow(image)
        else: axis.imshow(image, cmap='magma', vmin=0, vmax=1)
        if col == 3: axis.imshow(result['cam'], cmap='jet', alpha=0.43, vmin=0, vmax=1)
        axis.axis('off')
    axes[row_idx,0].set_title(f'Original: {CLASS_NAMES_10[local_target]}')
    axes[row_idx,1].set_title('Feature energy (class-agnostic)')
    axes[row_idx,2].set_title(f'CAM for {IMAGENET_CLASSES[target]}')
    axes[row_idx,3].set_title('CAM overlay')
fig.tight_layout(); fig.savefig(OUTPUT_DIR/'feature_visualization_cam.png', dpi=160, bbox_inches='tight')
plt.show()
````

**结果分析:**

## 4忠诚度评估

目的：如果XAI判断区域很重要，验证修改该区域是否确实显著影响模型对原预测类别的置信度。
我主要采用：

- Deletion：从原图出发，按 XAI 重要性从高到低依次遮蔽区域，记录固定目标类别概率,下降越快 / AUC 越低越好。
- Insertion：从基线图出发，按重要性从高到低逐步恢复区域，记录同一类别概率,上升越快 / AUC 越高越好。
- Random baseline：随机打乱块的顺序，使用相同遮蔽方式与相同模型。

### 公平性约定（组员共享）

1. 固定模型、权重、224×224 预处理、val 图像及样本顺序。
2. 固定目标类为未扰动原图的 1000 类 Top-1，且全过程不重新选类别。
3. 统一以14×14个不重叠patch排序（每块 16×16 像素），按 `steps=14` 等量扰动，避免只比较不同分辨率热力图的插值噪声。
4. 基线是归一化张量中的0，对应输入 RGB 空间的 ImageNet 三通道均值，不是纯黑 RGB。
5. `softmax` 概率在 `[0,1]`，曲线横轴为扰动图像的面积比例，使用 `np.trapezoid` / `np.trapz` 计算 AUC；**所有 XAI 方法共用相同样本和随机基线**。
6. 解释器必须返回与输入空间一致的非负重要性 `[H,W]`；对于 IG 等带符号方法，可先用 `attribution.clamp_min(0)`（正证据）或 `abs()`（绝对贡献），但需要在报告中写明并对所有比较结果说明含义。

````python
# 4.1 通用遮蔽 / Deletion / Insertion 计算

def as_nonnegative_map(saliency, height, width, signed_mode='positive'):
    """兼容 numpy / torch，输入 [H,W] 或 [1,H,W] 或 [1,1,H,W]。
    signed_mode='positive': 只看正贡献；'absolute': 看绝对贡献。
    """
    if isinstance(saliency, torch.Tensor):
        a = saliency.detach().cpu().float().numpy()
    else:
        a = np.asarray(saliency, dtype=np.float32)
    a = np.squeeze(a)
    if a.ndim != 2: raise ValueError(f'XAI 需要返回二维空间图，收到 {a.shape}')
    if not np.isfinite(a).all(): raise ValueError('XAI 图出现 NaN / Inf')
    if signed_mode == 'positive': a = np.maximum(a, 0)
    elif signed_mode == 'absolute': a = np.abs(a)
    else: raise ValueError('signed_mode 只能是 positive / absolute')
    t = torch.from_numpy(a.copy())[None,None]
    if a.shape != (height,width):
        t = F.interpolate(t, size=(height,width), mode='bilinear', align_corners=False)
    return t[0,0].numpy()


def pixel_order_to_patch_order(saliency, shape=(224,224), patch_grid=14, signed_mode='positive'):
    """重要性图 → patch 重要性降序排列，越前的 patch 越重要。"""
    h,w = shape
    assert h % patch_grid == 0 and w % patch_grid == 0, '输入宽高必须能被 patch_grid 整除'
    importance = as_nonnegative_map(saliency, h, w, signed_mode)
    patch_values = F.adaptive_avg_pool2d(torch.from_numpy(importance)[None,None],
                                         (patch_grid,patch_grid)).flatten().numpy()
    return np.argsort(-patch_values, kind='stable')


@torch.no_grad()
def perturbation_curve(model, input_tensor, target_class, ordered_patches,mode='deletion', steps=14, patch_grid=14):
    """根据给定 patch 顺序输出面积比例 vs 固定目标类 softmax 概率。
    0%: deletion=原图；insertion=全基线。100%: 两者互换。
    """
    if mode not in ('deletion','insertion'): raise ValueError('mode: deletion or insertion')
    assert input_tensor.ndim==4 and input_tensor.shape[0]==1
    h,w=input_tensor.shape[-2:]
    assert h%patch_grid==0 and w%patch_grid==0
    total=patch_grid**2
    order=np.asarray(ordered_patches, dtype=np.int64)
    assert len(order)==total and set(order.tolist())==set(range(total)), 'The patch order must be a sequence from 0 to grid²-1'
    rank=np.empty(total, dtype=np.int64)
    rank[order]=np.arange(total)
    rank=torch.from_numpy(rank).to(input_tensor.device).reshape(patch_grid,patch_grid)
    rank=rank.repeat_interleave(h//patch_grid,0).repeat_interleave(w//patch_grid,1)
    fractions=np.linspace(0,1,steps+1)
    levels=np.linspace(0,total,steps+1).round().astype(int)
    masks=torch.stack([(rank < int(k)) for k in levels])[:,None].float()  # [S,1,H,W]
    if mode=='deletion':
        variants=input_tensor*(1-masks)
    else:
        variants=input_tensor*masks
    batch_size=12 if input_tensor.device.type=='cpu' else 32
    scores=[]
    for batch in variants.split(batch_size,dim=0):
        scores.extend(model(batch).softmax(1)[:,int(target_class)].cpu().numpy().tolist())
    return fractions, np.array(scores, dtype=float)

def auc_of_curve(frac, conf):
    return float(np.trapezoid(conf,frac)) if hasattr(np,'trapezoid') else float(np.trapz(conf,frac))
````

````python
# 4.2 General Evaluator: Using the same batch of images, the same target category, and the same random baseline for all methods
# explainers is dict: method name -> fn(input_tensor[1,3,224,224], target:int) -> [224,224]

def evaluate_xai_methods(explainers, dataset, predictions, sample_indices,
                         steps=14, patch_grid=14, random_repeats=3, seed=2026,
                         signed_mode='positive'):
    if not explainers: raise ValueError('at least one XAI method')
    if len(sample_indices)==0: raise ValueError('at least one picture')
    pred_lookup=predictions.set_index('dataset_index')
    all_curves={k:{'deletion':[], 'insertion':[]} for k in explainers}
    all_curves['Random']={'deletion':[], 'insertion':[]}
    records=[]
    fractions=np.linspace(0,1,steps+1)
    for n,idx in enumerate(sample_indices,1):
        x,_=dataset[idx]; x=x[None].to(DEVICE)
        target=int(pred_lookup.loc[idx,'pred_imagenet1k_id'])
        h,w=x.shape[-2:]
        # The random baseline is shared across all methods to ensure a fair comparison.
        rng=np.random.default_rng(seed+int(idx))
        random_order_list=[rng.permutation(patch_grid**2) for _ in range(random_repeats)]
        random_pair=[]
        for order in random_order_list:
            random_pair.append({mode:perturbation_curve(MODEL,x,target,order,mode,steps,patch_grid)[1]for mode in ('deletion','insertion')})
        for mode in ('deletion','insertion'):
            random_avg=np.mean([p[mode] for p in random_pair],axis=0)
            all_curves['Random'][mode].append(random_avg)
            records.append({'dataset_index':idx,'method':'Random','mode':mode,'auc':auc_of_curve(fractions,random_avg)})
        for name,explainer in explainers.items():
            importance=explainer(x,target)
            order=pixel_order_to_patch_order(importance,shape=(h,w),patch_grid=patch_grid,signed_mode=signed_mode)
            for mode in ('deletion','insertion'):
                _, curve=perturbation_curve(MODEL,x,target,order,mode,steps,patch_grid)
                all_curves[name][mode].append(curve)
                records.append({'dataset_index':idx,'method':name,'mode':mode,'auc':auc_of_curve(fractions,curve)})
        print(f'Fidelity evaluated {n}/{len(sample_indices)}: idx={idx}, target={IMAGENET_CLASSES[target]}')
    curve_means={name:{mode:np.mean(arrs,axis=0) for mode,arrs in values.items()}for name,values in all_curves.items()}
    auc_df=pd.DataFrame(records)
    summary=auc_df.groupby(['method','mode'])['auc'].agg(['mean','std','count']).reset_index()
    return {'fractions':fractions, 'mean_curves':curve_means, 'sample_curves':all_curves,'per_image_auc':auc_df, 'summary':summary, 'indices':list(sample_indices)}

def plot_fidelity_comparison(evaluation, save_path=None):
    fig,axes=plt.subplots(1,2,figsize=(13,4.8))
    for ax,mode in zip(axes,['deletion','insertion']):
        for name,curve in evaluation['mean_curves'].items():
            style='--' if name=='Random' else '-'
            mean_auc=auc_of_curve(evaluation['fractions'],curve[mode])
            ax.plot(evaluation['fractions'],curve[mode],style,linewidth=2,
                    label=f'{name} (AUC={mean_auc:.3f})')
        ax.set(xlabel='Fraction of patches perturbed',ylabel='P(original predicted class)',
               xlim=(0,1),ylim=(0,1),title=f'{mode.title()} (lower AUC is better)' if mode=='deletion'
               else 'Insertion (higher AUC is better)')
        ax.grid(alpha=.2); ax.legend(fontsize=9)
    fig.suptitle(f'Faithfulness | N={len(evaluation["indices"])} images | same 1000-class model')
    fig.tight_layout()
    if save_path: fig.savefig(save_path, dpi=160, bbox_inches='tight')
    plt.show()
````

## 5 同一个 CNN 在不同条件下，关注区域是否稳定？

该演示不改变空间位置的亮度与模糊变换，使原图和变换图的 CAM 能在相同像素坐标中比较。这里用空间Pearson相关系数衡量热图一致性，范围为 `[-1,1]`，越接近 1 说明两张热图形状越相似。我们使用原图预测类别作为所有变换条件的固定解释目标；另列变换后预测是否发生变化。同时，所有变换都从同一张 224×224、经过中心裁剪的 RGB 原图出发，使用相同 ImageNet 归一化。

````python
# 5.1 Brightness / Blur without geometric distortion: small sample check
def change_conditions(x):
    rgb=torch.from_numpy(show_rgb(x)).permute(2,0,1)  # [3,H,W] in [0,1]
    mean=MEAN[0]; std=STD[0]
    original=x
    dark=(TF.adjust_brightness(rgb,0.70)[None]-mean)/std
    bright=(TF.adjust_brightness(rgb,1.30)[None]-mean)/std
    blurry=(TF.gaussian_blur(rgb,kernel_size=[9,9],sigma=[1.8,1.8])[None]-mean)/std
    return {'original': original,
            'brightness 0.7':dark.to(x.device),
            'brightness 1.3':bright.to(x.device),
            'gaussian blur':blurry.to(x.device)}
def spatial_pearson(a,b):
    a=np.asarray(a).ravel().astype(float); b=np.asarray(b).ravel().astype(float)
    if a.std()<1e-12 or b.std()<1e-12: return np.nan
    return float(np.corrcoef(a,b)[0,1])

condition_records=[]
idx=FIDELITY_INDICES[0]
x,_=VAL_DATASET[idx]; x=x[None].to(DEVICE)
original_target=int(PREDICTIONS.set_index('dataset_index').loc[idx,'pred_imagenet1k_id'])
variations=change_conditions(x)
original_map=fv_explainer(variations['original'],original_target)
fig, axes = plt.subplots(2,len(variations),figsize=(13,6))
for j,(cond,xcond) in enumerate(variations.items()):
    explain=FV.extract(xcond,original_target)
    new_pred=explain['prediction']
    correlation=spatial_pearson(original_map,explain['cam'])
    condition_records.append({'condition':cond,'cam_correlation':correlation,'pred_after':IMAGENET_CLASSES[new_pred],'pred_unchanged':new_pred==original_target})
    axes[0,j].imshow(show_rgb(xcond)); axes[0,j].axis('off'); axes[0,j].set_title(cond)
    axes[1,j].imshow(show_rgb(xcond)); axes[1,j].imshow(explain['cam'],cmap='jet',alpha=.43,vmin=0,vmax=1)
    axes[1,j].axis('off'); axes[1,j].set_title(f'CAM corr={correlation:.3f}\nlabel kept={new_pred==original_target}')
fig.suptitle('Feature Visualization under appearance changes (same fixed target)')
fig.tight_layout(); fig.savefig(OUTPUT_DIR/'feature_stability_conditions.png',dpi=160,bbox_inches='tight')
plt.show()
display(pd.DataFrame(condition_records))
````

**结果分析：**

在忠实性评估方面，本实验采用 Deletion和Insertion两种扰动方法，并将CAM与随机区域排序进行比较。实验共选择10张正确分类的图片，每个Imagenette类别一张。结果显示，CAM的DeletionAUC为0.1143，低于随机基线的0.1426；InsertionAUC为0.5725，明显高于随机基线的0.1253。
Deletion曲线显示，优先移除CAM 高亮区域能够更快降低目标类别的预测置信度,Insertion曲线则表明，优先恢复CAM高亮区域能够更快恢复模型的分类置信度。尤其是在仅恢复约36%图像区域时，CAM的平均预测置信度已达到约0.64，而随机恢复的置信度仍接近零。这表明CAM所识别的高重要性区域确实包含对模型分类具有较强影响的视觉信息。
此外，亮度扰动实验表明，CAM在所测试的Tench图片上保持了较高的空间相关性（0.9978和0.9990）。然而，在高斯模糊条件下，尽管CAM 相关系数仍达到0.9446，模型的预测类别却发生了变化。这说明关注区域的稳定性并不必然意味着预测结果稳定。
总体而言，Feature Visualization（CAM）能够提供具有一定可解释性且与模型预测行为相符的空间解释。在本次实验中，该方法的Deletion和Insertion结果均优于随机基线，尤其是在Insertion指标上表现突出。

**Feature Visualization的局限性**

Feature Visualization在这个研究中也有一定的局限性。首先，热力图空间分辨率较低,最后卷积层只有7×7的空间分辨率，之后通过插值放大到224×224，因此高亮区域的边界较模糊，不能视为精确的物体分割结果。其次，忠实性评估样本量有限,准确率是在全部3,925张验证图片上计算的，但忠实性仅在10张正确分类图片上评估，Insertion AUC 的样本标准差为 0.2290，说明不同图片之间存在明显差异，不能据此推断总体统计显著性。

# 7 Deep Feature Factorization (DFF) / 深度特征分解

**English:** DFF discovers recurring visual patterns in a CNN's internal features using nonnegative matrix factorization (NMF). Here we ask: which patterns recur across the existing Imagenette images, how do their loadings relate to the original class score, and how stable are they under brightness and blur changes? The ResNet18 classifier remains frozen. No new images, concept labels, or segmentation masks are required.

**中文：** DFF 使用非负矩阵分解（NMF）从 CNN 内部特征中发现重复出现的视觉模式。本节研究：已有 Imagenette 图片中有哪些共同模式？这些模式的激活与原类别分数有什么联系？亮度和模糊变化下它们是否稳定？ResNet18 分类器保持不变，无需收集新图片、概念标签或分割标注。

## 7.2 Method / 方法介绍

**English**

We extract the nonnegative, post-ReLU `layer4` activations of the existing ResNet18. Each 224 × 224 input yields a 512 × 7 × 7 tensor. Pooling spatial positions from up to 20 existing images per class gives a matrix $X\in\mathbb{R}_{+}^{(N\cdot7\cdot7)\times512}$. Class labels are used only to balance this subset, not as targets for NMF. All visualization and evaluation images are excluded from basis fitting.

NMF approximates $X\approx ZB$, where $B\in\mathbb{R}_{+}^{K\times512}$ contains shared feature directions and $Z\in\mathbb{R}_{+}^{(N\cdot7\cdot7)\times K}$ contains spatial loadings. We use $K=4$, NNDSVDa initialization, coordinate descent, Frobenius loss, seed 2026, and a maximum of 600 iterations. Each basis row is rescaled to unit L2 norm, with the corresponding loading rescaled inversely so the reconstruction is unchanged. The loadings become component maps. Components are named C1–C4, not assigned unverified semantic labels.

After fitting, **the same basis is kept fixed** for all new and transformed images; only their nonnegative loadings are solved. This keeps C1–C4 comparable across images and avoids confusing a permutation from independently refitted factorizations with a change in explanation. Per-component min-max scaling is used only for display. The mixture and loading shares use unnormalized loadings in the unit-norm basis; shares are not class probabilities.

**中文**

本实验提取现有 ResNet18 的 `layer4` 输出，即经过 ReLU 的非负特征。每张 224 × 224 输入对应 512 × 7 × 7 特征张量。默认从已有数据中每类选取至多 20 张图片，将空间位置合并成 $X\in\mathbb{R}_{+}^{(N\cdot7\cdot7)\times512}$。类别标签仅用于平衡抽样，不作为 NMF 的监督目标。所有展示图片和评估图片均不参与基底拟合。

NMF 将其近似分解为 $X\approx ZB$：$B\in\mathbb{R}_{+}^{K\times512}$ 表示共享特征方向，$Z\in\mathbb{R}_{+}^{(N\cdot7\cdot7)\times K}$ 表示各位置的激活系数。本实现取 $K=4$，使用 NNDSVDa 初始化、坐标下降、Frobenius 损失、随机种子 2026 和最多 600 次迭代。将各基底行归一化为单位 L2 范数，并同步调整激活系数，保持重建结果不变。激活系数恢复为空间热图，模式仅命名为 C1–C4，不自动断言它们对应某个真实物体或部件。

拟合完成后，**所有新图片和扰动图片都使用同一组固定基底**，只求解非负激活系数。这样才能跨图比较 C1–C4，避免将独立重新拟合造成的模式编号置换误认为解释变化。每张模式热图的 min-max 归一化仅用于显示；混合图和激活占比使用单位范数基底下的原始系数，占比不是类别概率。

**Connection to the class score / 与类别分数的联系**

The existing ResNet18 uses `layer4 → global average pooling → fc`. If $w_c$ and $b_c$ are the original classifier parameters and $\bar z_k$ is a component's mean spatial loading, then / 对于现有 ResNet18 的 `layer4 → 全局平均池化 → fc` 结构，设 $w_c,b_c$ 为原分类器参数，$\bar z_k$ 为模式的平均空间激活，则：

$$
z_c(x)\approx b_c+\sum_{k=1}^{K}\underbrace{(w_c^\top B_k)\bar z_k}_{\text{signed component logit contribution}}.
$$

This is a **derived linear-head decomposition**, not the definition of DFF or a calibrated concept probability. Its residual is reported explicitly. Raw DFF maps are class-agnostic, so they are not inserted into the original target-class Deletion/Insertion evaluator. We instead report feature reconstruction error, signed score decomposition, shared-pattern examples, and fixed-basis stability.

这是本实验附加的**线性分类头分解**，并非 DFF 本身的定义，也不是经过校准的概念概率。代码会明确报告近似分解的剩余误差。原始 DFF 热图不针对特定类别，因此不将其直接套入原有单目标类别的 Deletion/Insertion 评估；改为报告特征重建误差、有符号分数分解、共享模式样本及固定基底下的稳定性。

````python
# 7.3 Shared-basis DFF with temporary hooks and a frozen CNN.
class SharedBasisDFF:
    def __init__(self, model, layer, n_components=4, seed=2026, max_iter=600, tol=1e-4):
        self.model, self.layer = model, layer
        self.n_components = int(n_components)
        self.seed, self.max_iter, self.tol = int(seed), int(max_iter), float(tol)
        self.eps = 1e-8
        self.nmf = None
        self._cache = {}
        self.head_weights = model.fc.weight.detach().float().cpu().numpy().copy()
        self.head_bias = (model.fc.bias.detach().float().cpu().numpy().copy()
                          if model.fc.bias is not None else np.zeros(model.fc.out_features, dtype=np.float32))

    @torch.no_grad()
    def capture(self, inputs):
        # Record only the requested forward pass and always remove the hook.
        if self.model.training:
            raise RuntimeError('The CNN must remain in evaluation mode.')
        if inputs.ndim != 4 or inputs.device != next(self.model.parameters()).device:
            raise ValueError('Expected a batched image tensor on the model device.')
        values = []
        handle = self.layer.register_forward_hook(lambda module, args, output: values.append(output.detach().clone()))
        try:
            logits = self.model(inputs)
        finally:
            handle.remove()
        if len(values) != 1 or values[0].ndim != 4:
            raise RuntimeError('layer4 must produce exactly one [N,C,h,w] tensor per forward pass.')
        features = values[0].float().cpu().numpy()
        scores = logits.detach().float().cpu().numpy()
        if not np.isfinite(features).all() or not np.isfinite(scores).all():
            raise ValueError('Features or logits contain NaN or Inf.')
        if features.min() < -1e-6:
            raise ValueError('DFF requires nonnegative post-ReLU features; negative activations were found.')
        return np.maximum(features, 0), scores

    def _matrix(self, features):
        # Each row represents one spatial position, with channels as features.
        features = np.asarray(features, dtype=np.float32)
        if features.ndim != 4 or not np.isfinite(features).all() or features.min() < -1e-6:
            raise ValueError('Expected finite nonnegative [N,C,h,w] feature arrays.')
        n, c, h, w = features.shape
        matrix = np.maximum(features, 0).transpose(0, 2, 3, 1).reshape(n * h * w, c).copy()
        return matrix, (n, c, h, w)

    def fit(self, features):
        # Fit only NMF; this does not train or modify the CNN.
        matrix, (n, c, h, w) = self._matrix(features)
        if not 1 <= self.n_components <= min(matrix.shape) or np.linalg.norm(matrix) <= self.eps:
            raise ValueError('The requested rank is invalid or all basis-fitting activations are zero.')
        self.nmf = DFF_NMF(n_components=self.n_components, init='nndsvda', solver='cd',
                           beta_loss='frobenius', random_state=self.seed,
                           max_iter=self.max_iter, tol=self.tol)
        with DFF_warnings.catch_warnings(record=True) as caught:
            DFF_warnings.simplefilter('always', DFF_ConvergenceWarning)
            raw_loadings = self.nmf.fit_transform(matrix)
        original_basis = self.nmf.components_.astype(np.float32).copy()
        raw_norms = np.linalg.norm(original_basis, axis=1)
        self.scales = np.maximum(raw_norms, self.eps)
        self.basis = original_basis / self.scales[:, None]
        scaled_loadings = raw_loadings * self.scales[None, :]
        reconstruction = scaled_loadings @ self.basis
        self.fit_info = {
            'basis_images': n, 'channels': c, 'feature_height': h, 'feature_width': w,
            'components': self.n_components, 'iterations': int(self.nmf.n_iter_),
            'converged': not any(issubclass(warning.category, DFF_ConvergenceWarning) for warning in caught),
            'relative_feature_error': float(np.linalg.norm(matrix - reconstruction) / np.linalg.norm(matrix)),
            'inactive_components': int(np.count_nonzero(raw_norms <= self.eps)),
        }
        self._cache.clear()
        return scaled_loadings.reshape(n, h, w, self.n_components).transpose(0, 3, 1, 2)

    @staticmethod
    def display_maps(coefficients, height, width):
        # Normalize each map for display only, never for score decomposition.
        tensor = torch.from_numpy(np.asarray(coefficients, dtype=np.float32).copy())[None]
        up = F.interpolate(tensor, size=(height, width), mode='bilinear', align_corners=False)[0].numpy()
        low = up.min(axis=(1, 2), keepdims=True)
        span = up.max(axis=(1, 2), keepdims=True) - low
        return (up - low) / np.maximum(span, 1e-8), span[:, 0, 0] <= 1e-8

    @staticmethod
    def _copy(result):
        return {key: value.copy() if isinstance(value, np.ndarray) else value for key, value in result.items()}

    def extract(self, input_tensor, target_class):
        if self.nmf is None:
            raise RuntimeError('Fit the shared DFF basis first.')
        if input_tensor.ndim != 4 or input_tensor.shape[0] != 1 or self.model.training:
            raise ValueError('Expected one image and a CNN in evaluation mode.')
        target_class = int(target_class)
        if not 0 <= target_class < self.head_weights.shape[0]:
            raise ValueError('The fixed target class is outside the original model output range.')
        content = input_tensor.detach().to(device='cpu', dtype=torch.float32).contiguous().numpy().tobytes()
        key = (target_class, tuple(input_tensor.shape), DFF_hashlib.sha256(content).hexdigest())
        if key in self._cache:
            return self._copy(self._cache[key])
        features, logits = self.capture(input_tensor)
        matrix, (n, c, h, w) = self._matrix(features)
        if c != self.basis.shape[1] or c != self.head_weights.shape[1]:
            raise ValueError('Feature channels do not match the fitted basis and the original fc head.')
        # Keep the learned basis fixed; solve only the new image loadings.
        with DFF_warnings.catch_warnings(record=True) as caught:
            DFF_warnings.simplefilter('always', DFF_ConvergenceWarning)
            raw_loadings = self.nmf.transform(matrix)
        loadings = raw_loadings * self.scales[None, :]
        coefficients = loadings.reshape(h, w, self.n_components).transpose(2, 0, 1)
        reconstruction = loadings @ self.basis
        relative_error = float(np.linalg.norm(matrix - reconstruction) / max(np.linalg.norm(matrix), self.eps))
        strengths = coefficients.mean(axis=(1, 2))
        fractions = strengths / max(float(strengths.sum()), self.eps)
        maps, constant_maps = self.display_maps(coefficients, *input_tensor.shape[-2:])

        # Derive signed contributions through the known ResNet GAP -> linear head.
        association = self.basis @ self.head_weights[target_class]
        contributions = association * strengths
        bias = float(self.head_bias[target_class])
        original_logit = float(logits[0, target_class])
        checked_logit = float(features[0].mean(axis=(1, 2)) @ self.head_weights[target_class] + bias)
        if not np.isclose(original_logit, checked_logit, rtol=1e-4, atol=1e-4):
            raise RuntimeError('The model no longer matches layer4 -> global average pooling -> fc.')
        reconstructed_logit = float(bias + contributions.sum())
        shifted = logits[0] - logits[0].max()
        probabilities = np.exp(shifted) / np.exp(shifted).sum()
        result = {
            'maps': maps.astype(np.float32), 'coefficients': coefficients.astype(np.float32),
            'component_strengths': strengths, 'loading_fractions': fractions,
            'head_association': association, 'logit_contributions': contributions,
            'feature_relative_error': relative_error, 'target_class': target_class,
            'target_probability': float(probabilities[target_class]),
            'prediction': int(logits[0].argmax()), 'original_logit': original_logit,
            'head_bias': bias, 'reconstructed_logit': reconstructed_logit,
            'logit_residual': original_logit - reconstructed_logit,
            'constant_maps': constant_maps,
            'projection_converged': not any(issubclass(warning.category, DFF_ConvergenceWarning) for warning in caught),
        }
        if len(self._cache) >= 128:
            self._cache.pop(next(iter(self._cache)))
        self._cache[key] = self._copy(result)
        return result


DFF_ENGINE = SharedBasisDFF(MODEL, MODEL.layer4, DFF_N_COMPONENTS, SEED, DFF_MAX_ITER, DFF_TOL)
````

## 7.5 Component maps and recurring examples / 模式热图与重复出现的样本

**English:** C1–C4 have the same basis definitions in every image. The mixed-color view shows relative component loadings; each separate heatmap is normalized independently for visibility. The exemplar grid shows the three basis-fitting images with the highest mean loading for each component. These are examples for human inspection, not ground-truth concept names or segmentation masks.

**中文：** 不同图片中的 C1–C4 始终对应同一组基底。混合颜色图展示模式的相对激活，单独模式热图则分别归一化以便观察。代表样本网格展示每个模式在基底拟合集内平均激活最高的三张图片，用于人工检查共同结构，不代表真实概念标签或分割标注。

````python
def dff_mixture(rgb, coefficients):
    # Mix component colors using raw loadings in the unit-norm basis.
    tensor = torch.from_numpy(np.asarray(coefficients, dtype=np.float32).copy())[None]
    up = F.interpolate(tensor, size=rgb.shape[:2], mode='bilinear', align_corners=False)[0].numpy()
    total = up.sum(axis=0)
    relative = up / np.maximum(total[None], 1e-8)
    colors = np.einsum('khw,kc->hwc', relative, DFF_COLORS)
    return np.where((total > 1e-8)[..., None], 0.6 * rgb + 0.4 * colors, rgb).clip(0, 1)


# Display the same fixed visualization images used in the earlier experiments.
DFF_VIS_FIG, DFF_VIS_AXES = plt.subplots(len(DFF_VIS_INDICES), 2 + DFF_N_COMPONENTS,
    figsize=(3 * (2 + DFF_N_COMPONENTS), 3.2 * len(DFF_VIS_INDICES)), squeeze=False)
DFF_VIS_DIAGNOSTICS = []
for DFF_row, DFF_idx in enumerate(DFF_VIS_INDICES):
    DFF_x, _ = VAL_DATASET[DFF_idx]
    DFF_x = DFF_x[None].to(DEVICE)
    DFF_target = int(DFF_PRED_LOOKUP.loc[DFF_idx, 'pred_imagenet1k_id'])
    DFF_result = DFF_ENGINE.extract(DFF_x, DFF_target)
    DFF_rgb = show_rgb(DFF_x)
    DFF_VIS_AXES[DFF_row, 0].imshow(DFF_rgb)
    DFF_VIS_AXES[DFF_row, 0].set_title(f'Original: {IMAGENET_CLASSES[DFF_target]}', fontsize=9)
    DFF_VIS_AXES[DFF_row, 1].imshow(dff_mixture(DFF_rgb, DFF_result['coefficients']))
    DFF_VIS_AXES[DFF_row, 1].set_title('Shared component mixture')
    for DFF_component in range(DFF_N_COMPONENTS):
        DFF_axis = DFF_VIS_AXES[DFF_row, 2 + DFF_component]
        DFF_axis.imshow(DFF_rgb)
        DFF_axis.imshow(DFF_result['maps'][DFF_component], cmap='magma', alpha=0.5, vmin=0, vmax=1)
        DFF_axis.set_title(DFF_COMPONENT_NAMES[DFF_component])
    for DFF_axis in DFF_VIS_AXES[DFF_row]:
        DFF_axis.axis('off')
    DFF_VIS_DIAGNOSTICS.append({'dataset_index': DFF_idx,
        'feature_relative_error': DFF_result['feature_relative_error'],
        'constant_maps': int(DFF_result['constant_maps'].sum()),
        'projection_converged': DFF_result['projection_converged']})
    np.savez_compressed(DFF_OUTPUT_DIR / f'dff_image_{DFF_idx}.npz',
                        component_maps=DFF_result['maps'], coefficients=DFF_result['coefficients'],
                        logit_contributions=DFF_result['logit_contributions'], target_class=DFF_target)
DFF_VIS_FIG.legend(handles=[DFF_Patch(color=DFF_COLORS[k], label=DFF_COMPONENT_NAMES[k])
                           for k in range(DFF_N_COMPONENTS)], loc='upper center', ncol=DFF_N_COMPONENTS)
DFF_VIS_FIG.tight_layout(rect=(0, 0, 1, 0.98))
DFF_VIS_FIG.savefig(DFF_OUTPUT_DIR / 'dff_shared_component_maps.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(DFF_VIS_FIG)
pd.DataFrame(DFF_VIS_DIAGNOSTICS).to_csv(DFF_OUTPUT_DIR / 'dff_visualization_diagnostics.csv', index=False)

# Retrieve recurring image examples using fitting-set loadings, without another CNN pass.
DFF_EXAMPLE_COUNT = min(3, len(DFF_FIT_INDICES))
DFF_EXAMPLE_FIG, DFF_EXAMPLE_AXES = plt.subplots(DFF_N_COMPONENTS, DFF_EXAMPLE_COUNT,
    figsize=(4 * DFF_EXAMPLE_COUNT, 3.3 * DFF_N_COMPONENTS), squeeze=False)
DFF_FIT_STRENGTHS = DFF_FIT_COEFFICIENTS.mean(axis=(2, 3))
DFF_EXEMPLAR_ROWS = []
for DFF_component in range(DFF_N_COMPONENTS):
    DFF_positions = np.argsort(-DFF_FIT_STRENGTHS[:, DFF_component], kind='stable')[:DFF_EXAMPLE_COUNT]
    for DFF_rank, DFF_position in enumerate(DFF_positions):
        DFF_idx = DFF_FIT_INDICES[int(DFF_position)]
        DFF_x, _ = VAL_DATASET[DFF_idx]
        DFF_rgb = show_rgb(DFF_x[None])
        DFF_maps, _ = DFF_ENGINE.display_maps(DFF_FIT_COEFFICIENTS[DFF_position], *DFF_rgb.shape[:2])
        DFF_axis = DFF_EXAMPLE_AXES[DFF_component, DFF_rank]
        DFF_axis.imshow(DFF_rgb)
        DFF_axis.imshow(DFF_maps[DFF_component], cmap='magma', alpha=0.5, vmin=0, vmax=1)
        DFF_pred = int(DFF_FIT_LOGITS[DFF_position].argmax())
        DFF_axis.set_title(f'{DFF_COMPONENT_NAMES[DFF_component]} | rank {DFF_rank + 1} | idx={DFF_idx}\n'
                           f'Prediction: {IMAGENET_CLASSES[DFF_pred]}', fontsize=9)
        DFF_axis.axis('off')
        DFF_EXEMPLAR_ROWS.append({'component': DFF_COMPONENT_NAMES[DFF_component], 'rank': DFF_rank + 1,
            'dataset_index': DFF_idx, 'prediction': IMAGENET_CLASSES[DFF_pred],
            'mean_loading': float(DFF_FIT_STRENGTHS[DFF_position, DFF_component])})
DFF_EXAMPLE_FIG.tight_layout()
DFF_EXAMPLE_FIG.savefig(DFF_OUTPUT_DIR / 'dff_component_exemplars.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(DFF_EXAMPLE_FIG)
pd.DataFrame(DFF_EXEMPLAR_ROWS).to_csv(DFF_OUTPUT_DIR / 'dff_component_exemplars.csv', index=False)
````

## 7.6 Shared loading profiles and class-score decomposition / 共享激活与类别分数分解

**English:** Evaluate the original fixed images on the learned basis. Loading fractions describe how activity is distributed among the normalized basis components; they are not importance percentages. Signed component contributions approximate the fixed target's logit through the original classifier, with the classifier bias and reconstruction residual shown separately. A bright component map can have either a positive or a negative association with a particular class.

**中文：** 将原来的固定评估图片投影到已学习的基底上。激活占比描述单位范数基底下各模式的相对活动，不是重要性百分比。有符号模式贡献通过原分类器近似分解固定目标类别的 logit，并另外报告分类器偏置和重建剩余误差。同一个高亮模式对不同类别可能具有正向或负向联系。

````python
DFF_EVAL_ROWS, DFF_COMPONENT_ROWS = [], []
for DFF_idx in DFF_EVAL_INDICES:
    DFF_x, _ = VAL_DATASET[DFF_idx]
    DFF_target = int(DFF_PRED_LOOKUP.loc[DFF_idx, 'pred_imagenet1k_id'])
    DFF_result = DFF_ENGINE.extract(DFF_x[None].to(DEVICE), DFF_target)
    DFF_EVAL_ROWS.append({'dataset_index': DFF_idx, 'target_class': IMAGENET_CLASSES[DFF_target],
        **{key: DFF_result[key] for key in ('feature_relative_error', 'original_logit', 'head_bias',
            'reconstructed_logit', 'logit_residual', 'projection_converged')},
        'constant_maps': int(DFF_result['constant_maps'].sum())})
    for DFF_component, DFF_name in enumerate(DFF_COMPONENT_NAMES):
        DFF_COMPONENT_ROWS.append({'dataset_index': DFF_idx, 'target_class': IMAGENET_CLASSES[DFF_target],
            'component': DFF_name, 'mean_loading': float(DFF_result['component_strengths'][DFF_component]),
            'loading_fraction': float(DFF_result['loading_fractions'][DFF_component]),
            'linear_head_association': float(DFF_result['head_association'][DFF_component]),
            'signed_logit_contribution': float(DFF_result['logit_contributions'][DFF_component])})
DFF_EVAL_SUMMARY = pd.DataFrame(DFF_EVAL_ROWS)
DFF_COMPONENT_PROFILES = pd.DataFrame(DFF_COMPONENT_ROWS)
DFF_EVAL_SUMMARY.to_csv(DFF_OUTPUT_DIR / 'dff_evaluation_reconstruction.csv', index=False)
DFF_COMPONENT_PROFILES.to_csv(DFF_OUTPUT_DIR / 'dff_component_profiles.csv', index=False)
DFF_display(DFF_EVAL_SUMMARY.round(4))
DFF_display(DFF_COMPONENT_PROFILES.round(4))

# The fixed component order is shared across all images and both summary plots.
DFF_LOAD_TABLE = DFF_COMPONENT_PROFILES.pivot(index='dataset_index', columns='component',
    values='loading_fraction').reindex(index=list(DFF_EVAL_INDICES), columns=DFF_COMPONENT_NAMES)
DFF_CONTRIB_TABLE = DFF_COMPONENT_PROFILES.pivot(index='dataset_index', columns='component',
    values='signed_logit_contribution').reindex(index=list(DFF_EVAL_INDICES), columns=DFF_COMPONENT_NAMES)
DFF_PROFILE_FIG, DFF_PROFILE_AXES = plt.subplots(1, 2, figsize=(13, 6))
DFF_PROFILE_IMAGE = DFF_PROFILE_AXES[0].imshow(DFF_LOAD_TABLE.to_numpy(), cmap='viridis', vmin=0, vmax=1, aspect='auto')
DFF_SCALE = max(float(np.abs(DFF_CONTRIB_TABLE.to_numpy()).max()), 1e-8)
DFF_CONTRIB_IMAGE = DFF_PROFILE_AXES[1].imshow(DFF_CONTRIB_TABLE.to_numpy(), cmap='coolwarm',
    vmin=-DFF_SCALE, vmax=DFF_SCALE, aspect='auto')
for DFF_axis in DFF_PROFILE_AXES:
    DFF_axis.set_xticks(range(DFF_N_COMPONENTS), DFF_COMPONENT_NAMES)
    DFF_axis.set_yticks(range(len(DFF_EVAL_INDICES)),
        [f'{DFF_EVAL_ROWS[i]["target_class"]} ({idx})' for i, idx in enumerate(DFF_EVAL_INDICES)])
DFF_PROFILE_AXES[0].set_title('Relative component loading (not probability)')
DFF_PROFILE_AXES[1].set_title('Signed target-logit contribution (approximation)')
DFF_PROFILE_FIG.colorbar(DFF_PROFILE_IMAGE, ax=DFF_PROFILE_AXES[0], shrink=0.7)
DFF_PROFILE_FIG.colorbar(DFF_CONTRIB_IMAGE, ax=DFF_PROFILE_AXES[1], shrink=0.7)
DFF_PROFILE_FIG.tight_layout()
DFF_PROFILE_FIG.savefig(DFF_OUTPUT_DIR / 'dff_component_profiles.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(DFF_PROFILE_FIG)
````

## 7.7 Fixed-basis stability / 固定基底下的稳定性

**English:** The basis is not refitted after brightness or blur changes. Each transformed C1 map is compared with the original C1 map, and similarly for all other components. Pearson correlation measures spatial map similarity; loading-profile L1 distance measures redistribution across components. Constant maps have undefined correlation and are counted explicitly. The model's prediction retention is recorded separately. These component-level metrics are not directly interchangeable with the earlier target-class CAM fidelity scores.

**中文：** 亮度或模糊变化后不重新拟合基底，而是将变化后的 C1 与原图 C1 比较，其余模式同理。Pearson 相关系数衡量空间热图相似性，激活占比向量的 L1 距离衡量模式之间的活动重新分配。常量热图的相关系数没有定义，代码会明确记录有效数量；模型预测是否保持则另外统计。这些模式层面的指标不能直接替代前面的目标类别 CAM 忠实性分数。

````python
DFF_STABILITY_ROWS, DFF_CONDITION_ROWS = [], []
for DFF_image_number, DFF_idx in enumerate(DFF_EVAL_INDICES):
    DFF_x, _ = VAL_DATASET[DFF_idx]
    DFF_x = DFF_x[None].to(DEVICE)
    DFF_target = int(DFF_PRED_LOOKUP.loc[DFF_idx, 'pred_imagenet1k_id'])
    DFF_reference = DFF_ENGINE.extract(DFF_x, DFF_target)
    DFF_variants = change_conditions(DFF_x)
    if DFF_image_number == 0:
        DFF_STABILITY_FIG, DFF_STABILITY_AXES = plt.subplots(1 + DFF_N_COMPONENTS, len(DFF_variants),
            figsize=(14, 2.7 * (1 + DFF_N_COMPONENTS)), squeeze=False)
    for DFF_column, (DFF_condition, DFF_variant) in enumerate(DFF_variants.items()):
        # Solve loadings using the same basis even when the classifier changes its prediction.
        DFF_changed = DFF_reference if DFF_condition == 'original' else DFF_ENGINE.extract(DFF_variant, DFF_target)
        DFF_kept = DFF_changed['prediction'] == DFF_target
        DFF_profile_l1 = float(np.abs(DFF_changed['loading_fractions'] - DFF_reference['loading_fractions']).sum())
        DFF_CONDITION_ROWS.append({'dataset_index': DFF_idx, 'condition': DFF_condition,
            'loading_profile_l1': DFF_profile_l1, 'prediction_kept': DFF_kept,
            'fixed_target': IMAGENET_CLASSES[DFF_target],
            'prediction_after': IMAGENET_CLASSES[DFF_changed['prediction']],
            'target_probability': DFF_changed['target_probability'],
            'feature_relative_error': DFF_changed['feature_relative_error'],
            'projection_converged': DFF_changed['projection_converged']})
        if DFF_image_number == 0:
            DFF_STABILITY_AXES[0, DFF_column].imshow(show_rgb(DFF_variant))
            DFF_STABILITY_AXES[0, DFF_column].set_title(
                f'{DFF_condition}\nP(target)={DFF_changed["target_probability"]:.3f}; label kept={DFF_kept}', fontsize=9)
            DFF_STABILITY_AXES[0, DFF_column].axis('off')
        for DFF_component, DFF_name in enumerate(DFF_COMPONENT_NAMES):
            DFF_corr = spatial_pearson(DFF_reference['maps'][DFF_component], DFF_changed['maps'][DFF_component])
            DFF_STABILITY_ROWS.append({'dataset_index': DFF_idx, 'condition': DFF_condition,
                'component': DFF_name, 'map_correlation': DFF_corr,
                'absolute_loading_fraction_change': float(abs(DFF_changed['loading_fractions'][DFF_component]
                                                             - DFF_reference['loading_fractions'][DFF_component]))})
            if DFF_image_number == 0:
                DFF_axis = DFF_STABILITY_AXES[DFF_component + 1, DFF_column]
                DFF_axis.imshow(show_rgb(DFF_variant))
                DFF_axis.imshow(DFF_changed['maps'][DFF_component], cmap='magma', alpha=0.5, vmin=0, vmax=1)
                DFF_axis.set_title(f'{DFF_name}: corr={DFF_corr:.3f}', fontsize=9)
                DFF_axis.axis('off')
    print(f'DFF stability {DFF_image_number + 1}/{len(DFF_EVAL_INDICES)}: idx={DFF_idx}')
DFF_STABILITY_FIG.suptitle('Shared DFF basis under appearance changes | first evaluation image')
DFF_STABILITY_FIG.tight_layout()
DFF_STABILITY_FIG.savefig(DFF_OUTPUT_DIR / 'dff_stability_example.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(DFF_STABILITY_FIG)

# Count valid correlations separately so undefined values are not interpreted as perfect stability.
DFF_STABILITY = pd.DataFrame(DFF_STABILITY_ROWS)
DFF_CONDITIONS = pd.DataFrame(DFF_CONDITION_ROWS)
DFF_STABILITY_SUMMARY = DFF_STABILITY.groupby(['condition', 'component'], sort=False).agg(
    mean_correlation=('map_correlation', 'mean'), std_correlation=('map_correlation', 'std'),
    valid_correlations=('map_correlation', 'count'), n_images=('dataset_index', 'size'),
    mean_loading_fraction_change=('absolute_loading_fraction_change', 'mean')).reset_index()
DFF_CONDITION_SUMMARY = DFF_CONDITIONS.groupby('condition', sort=False).agg(
    mean_loading_profile_l1=('loading_profile_l1', 'mean'),
    prediction_kept_rate=('prediction_kept', 'mean'),
    mean_feature_error=('feature_relative_error', 'mean'),
    n_images=('dataset_index', 'size')).reset_index()
for DFF_name, DFF_table in [('stability_per_component', DFF_STABILITY), ('conditions_per_image', DFF_CONDITIONS),
                            ('stability_summary', DFF_STABILITY_SUMMARY), ('condition_summary', DFF_CONDITION_SUMMARY)]:
    DFF_table.to_csv(DFF_OUTPUT_DIR / f'dff_{DFF_name}.csv', index=False)
DFF_display(DFF_STABILITY_SUMMARY.round(4))
DFF_display(DFF_CONDITION_SUMMARY.round(4))
````

**中文：** 上述百分比只是单位范数基底下的相对 NMF 激活系数，并非类别概率或因果重要性；分类头关联数值也是线性分数，并非经过校准的置信度。各热图为了显示而分别归一化，因此颜色较亮的模式也可能激活很弱，例如 C3 在鱼图片中的激活占比仅为 1.93%。这些图支持探索性的模式描述，不能直接当作已经验证的物体部件标签或精确分割边界。

### 7.8.3 Connection to the original class score / 与原类别分数的联系

Selected rows from the saved reconstruction table are shown below. Logits are unnormalized model scores, not probabilities. / 下表摘录已保存的重建结果；logit 是未归一化的模型分数，不是概率。

| Fixed target / 固定目标 | Original logit / 原始 logit | Reconstructed logit / 重建 logit | Residual / 剩余误差 |
|---|---:|---:|---:|
| tench | 11.8568 | 11.2856 | 0.5712 |
| golf ball | 19.9688 | 16.8143 | 3.1545 |
| parachute | 23.3014 | 17.5362 | 5.7652 |
| cassette player | 18.3643 | 5.8627 | 12.5016 |
| English springer | 13.1185 | 2.1500 | 10.9685 |
| church | 10.6936 | 0.1922 | 10.5014 |
| gas pump | 21.3854 | 3.4186 | 17.9668 |

**English:** The four-factor approximation captures the selected tench score much more closely than several other class scores. For tench, C2 contributes +14.2022 logit units, while C1, C3, and C4 contribute −0.5443, −0.0515, and −2.3182. Combining these signed terms with the classifier bias gives the approximate logit 11.2856; the residual to the original score is 0.5712. This shows why an individual component contribution can exceed the final logit and must not be read as a percentage. In contrast, the gas-pump residual is 17.9668, and the mean absolute residual across all 10 images is 9.4821. A recognizable or dominant component therefore does not imply that the original decision has been fully explained. Considerable class-relevant information remains outside the chosen rank-four reconstruction.

**中文：** 四模式近似对选定鱼图片的类别分数重建较接近，但对若干其他类别的覆盖明显不足。鱼图片中，C2 贡献 +14.2022 个 logit 单位，C1、C3、C4 分别贡献 −0.5443、−0.0515、−2.3182；这些有符号项与分类器偏置结合后得到近似 logit 11.2856，与原分数相差 0.5712。这说明单个模式贡献可以大于最终 logit，不能将其理解为贡献百分比。相比之下，加油机图片的剩余误差达到 17.9668，全部 10 张图片的平均绝对剩余误差为 9.4821。因此，即使某模式容易辨认或激活占比较高，也不意味着原决策已得到完整解释；当前四阶重建之外仍保留了相当多与类别有关的信息。

### 7.8.4 Stability under appearance changes / 外观变化下的稳定性

The same fitted basis is used in every condition. Mean component correlation averages 10 images × 4 component maps, while prediction retention is calculated over 10 images. The 40 maps are not 40 independent image samples. / 所有条件都使用同一组已拟合基底。模式平均相关系数汇总 10 张图片 × 4 个模式热图，预测保持率则以 10 张图片为单位；40 张模式热图并非 40 个独立图片样本。

| Condition / 条件 | Mean component correlation / 模式平均相关系数 | Loading-profile L1 / 激活占比 L1 | Prediction retained / 预测保持率 |
|---|---:|---:|---:|
| Original / 原图 | 1.0000 | 0.0000 | 100% |
| Brightness × 0.7 / 亮度 × 0.7 | 0.9601 | 0.0589 | 90% |
| Brightness × 1.3 / 亮度 × 1.3 | 0.9436 | 0.0433 | 100% |
| Gaussian blur / 高斯模糊 | 0.7935 | 0.1287 | 70% |

**English:** Brightness changes preserve the component maps relatively well, with mean spatial correlations above 0.94 and limited redistribution of component loading. Blur produces a larger change in both spatial layout and loading composition: the mean correlation falls to 0.7935 and profile L1 distance rises to 0.1287. The mean feature-reconstruction error also increases from 0.7680 to 0.7931, so the fixed original-image basis approximates blurred features less well on average. Spatial correlation and loading-profile distance measure different properties; the two brightness conditions do not have the same ordering on both metrics.

**中文：** 亮度变化下，模式热图总体较稳定，平均空间相关系数均高于 0.94，模式激活分配的变化也较小。模糊同时引起更明显的空间变化与模式构成变化：平均相关系数降至 0.7935，激活占比向量的 L1 距离升至 0.1287。平均特征重建误差也从 0.7680 增加到 0.7931，说明原图拟合的固定基底平均而言较难近似模糊后的特征。空间相关性与激活分布距离反映不同性质，因此两个亮度条件在这两项指标上的排序并不完全一致。

**English:** Stability is also component-dependent. Under blur, C4 has the highest mean correlation (0.9114), followed by C1 (0.8196), C3 (0.7669), and C2 (0.6764). C2 has a large across-image standard deviation of 0.3704, so its behavior is heterogeneous. This does not establish that fish-related evidence always disappears under blur: most evaluated images are not fish, and independently normalized maps with weak loading can be sensitive to small changes. Active-component or per-class analysis would be needed to isolate that claim.

**中文：** 不同模式的稳定性也不同。模糊条件下，C4 的平均相关系数最高（0.9114），其次是 C1（0.8196）、C3（0.7669）和 C2（0.6764）。C2 的跨图标准差达到 0.3704，说明个体差异较大。这不能证明鱼类相关证据在模糊后总会消失：大部分评估图片并非鱼，而且激活较弱的热图经过独立归一化后，也可能对小变化较敏感。若要验证这种语义层面的结论，需要进一步按类别或高激活模式分析。

### 7.8.5 Map stability versus prediction stability / 热图稳定与预测稳定

**English:** In the first tench example, blur lowers the fixed target probability from approximately 0.592 to 0.032 and changes the predicted class, although the fish-associated C2 map still has correlation 0.942 with its original map. This is an important counterexample to treating a stable heatmap as evidence of a stable decision. Pearson correlation describes map shape and is insensitive to positive rescaling; it does not require unchanged activation magnitude, class-score contributions, competing-class scores, or reconstruction residuals. Across the sample, the CNN retains its class for 9/10 darkened, 10/10 brightened, and 7/10 blurred images. These rates describe the same underlying CNN, not a classification accuracy achieved by DFF.

**中文：** 第一张鱼图片中，模糊使固定目标类别概率从约 0.592 降到 0.032，并改变了预测类别，但鱼相关模式 C2 与原图热图的相关系数仍为 0.942。这直接说明热图稳定并不保证决策稳定。Pearson 相关系数描述热图形状，对正比例缩放不敏感，因而不要求激活幅度、类别分数贡献、竞争类别分数或重建剩余信息保持不变。整组样本中，变暗、变亮及模糊后的类别保持数量分别为 9/10、10/10、7/10；这些保持率描述的是同一个 CNN，而不是 DFF 所获得的分类准确率。

### 7.8.6 Research conclusion and limits / 研究结论与局限

**English:** DFF adds a representation-level view to the earlier CAM analysis. It reveals reusable feature directions with tentative fish, ball, parachute, and broader device-related associations, and shows how their activity redistributes under perturbation. However, the current four-component model has substantial feature and logit reconstruction residuals, so it provides a partial summary rather than a complete account of classification. The clearest robustness observation is stronger sensitivity to blur than to the tested brightness changes, together with component-specific variability and a separation between map and prediction stability.

**中文：** DFF 为前面的 CAM 分析补充了表示层面的视角：它揭示了可复用的特征方向，与鱼类、球类、降落伞及更宽泛的设备模式具有暂时性的关联，并显示了扰动下模式活动的重新分配。但当前四模式模型存在较大的特征和 logit 重建剩余误差，因此它提供的是部分概括，而非分类过程的完整说明。稳定性方面最明确的观察是：相较于本次亮度变化，模糊造成的影响更大；不同模式的敏感程度不同，而且热图稳定与预测稳定需要分别考察。

**English:** These observations are limited to 200 basis-fitting images, one fitted rank/seed configuration, and 10 selected correct evaluation examples. The 7 × 7 source resolution and absence of concept/segmentation ground truth prevent claims of precise localization or verified concept accuracy.
