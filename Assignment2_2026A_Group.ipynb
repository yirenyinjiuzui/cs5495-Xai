````python
from pathlib import Path
import os, time, random, tarfile, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset
from torchvision import models, datasets, transforms
from torchvision.transforms import functional as TF
from PIL import Image, ImageEnhance, ImageFilter

# if use Google Drive
# from google.colab import drive
# drive.mount('/content/drive')

BASE_DIR = Path('/content') if Path('/content').is_dir() else Path('/mnt/data')
ARCHIVE_PATH = BASE_DIR / 'imagenette2-320.tgz'  #  Drive Path (if want)
WEIGHT_PATH  = BASE_DIR / 'resnet18-f37072fd.pth'
DATA_ROOT    = BASE_DIR / 'imagenette2-320'
OUTPUT_DIR   = BASE_DIR / 'xai_outputs'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
QUICK_TEST = False
BATCH_SIZE = 64 if torch.cuda.is_available() else 32
NUM_WORKERS = 2
SEED = 2026
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
if DEVICE.type == 'cpu':
    torch.set_num_threads(min(4, os.cpu_count() or 1))
print(f'Device={DEVICE}, PyTorch={torch.__version__}, output={OUTPUT_DIR}')
print(f'Weight found: {WEIGHT_PATH.is_file()}; archive found: {ARCHIVE_PATH.is_file()}')
assert WEIGHT_PATH.is_file(), f'Can not find the model weights, please check WEIGHT_PATH: {WEIGHT_PATH}'
assert ARCHIVE_PATH.is_file() or (DATA_ROOT/'val').is_dir(), f'Can not find the dataset, please check ARCHIVE_PATH: {ARCHIVE_PATH}'
````

## 1 解压数据与模型预处理
在实验中数据使用 `val`，评估泛化性能，使用ResNet18官方 ImageNet-1K V1预处理：短边 Resize 到 256 → CenterCrop 224 → ToTensor → ImageNet 均值/标准差归一化。类别 ID 映射是必要步骤：`ImageFolder` 的 10 类本地标签（0–9）不能与 ResNet18 1000 类输出直接对比。

````python
if not (DATA_ROOT/'val').is_dir():
    print('Extracting', ARCHIVE_PATH.name)
    with tarfile.open(ARCHIVE_PATH, 'r:gz') as tar:
        members = tar.getmembers()
        for m in members:
            p = Path(m.name)
            if p.is_absolute() or '..' in p.parts or not (m.isfile() or m.isdir()):
                raise ValueError(f'The zip file contains unsafe entries: {m.name}')
        try:
            tar.extractall(path=BASE_DIR, members=members, filter='data')
        except TypeError:
            tar.extractall(path=BASE_DIR, members=members)
print('Validation set directory:', DATA_ROOT/'val')

WEIGHTS_META = models.ResNet18_Weights.IMAGENET1K_V1
TRANSFORM = WEIGHTS_META.transforms()
IMAGENET_CLASSES = WEIGHTS_META.meta['categories']

# The order is the same as the WNID alphabetical order in ImageFolder, and the assertion prevents data labels from silently misaligning.
WNID_TO_NAME = {
    'n01440764': 'tench',
    'n02102040': 'English springer',
    'n02979186': 'cassette player',
    'n03000684': 'chain saw',
    'n03028079': 'church',
    'n03394916': 'French horn',
    'n03417042': 'garbage truck',
    'n03425413': 'gas pump',
    'n03445777': 'golf ball',
    'n03888257': 'parachute',}
VAL_DATASET = datasets.ImageFolder(str(DATA_ROOT/'val'), transform=TRANSFORM)
assert set(VAL_DATASET.classes) == set(WNID_TO_NAME), 'The directory categories do not match the official Imagenette categories'
CLASS_NAMES_10 = [WNID_TO_NAME[k] for k in VAL_DATASET.classes]
IMAGENET_ID_BY_LOCAL = [IMAGENET_CLASSES.index(n) for n in CLASS_NAMES_10]
assert IMAGENET_ID_BY_LOCAL == [0, 217, 482, 491, 497, 566, 569, 571, 574, 701]
assert len(VAL_DATASET) == 3925, f' output {len(VAL_DATASET)} '
print('Full validation set：', len(VAL_DATASET), '；ImageNet output 1000 class')
print(pd.DataFrame({'WNID': VAL_DATASET.classes, 'Imagenette label': range(10),
                    'ImageNet-1K label': IMAGENET_ID_BY_LOCAL, 'class': CLASS_NAMES_10}).to_string(index=False))

MODEL = models.resnet18(weights=None)
try:
    state = torch.load(WEIGHT_PATH, map_location='cpu', weights_only=True)
except TypeError:
    state = torch.load(WEIGHT_PATH, map_location='cpu')
MODEL.load_state_dict(state, strict=True)
MODEL = MODEL.to(DEVICE).eval()
assert MODEL.fc.out_features == 1000
print('Model loaded successfully, output dimensions：', MODEL.fc.out_features)
````

## 2完整验证集预测与模型准确率


````python
# 2.1 Predict 1000 classes
EVAL_DATASET = Subset(VAL_DATASET, range(min(256,len(VAL_DATASET)))) if QUICK_TEST else VAL_DATASET
EVAL_INDICES = list(range(len(EVAL_DATASET)))
loader = DataLoader(EVAL_DATASET, batch_size=BATCH_SIZE, shuffle=False,num_workers=NUM_WORKERS, pin_memory=(DEVICE.type=='cuda'))
rows = []; started = time.time()
with torch.inference_mode():
    current = 0
    for images, local_targets in loader:
        images = images.to(DEVICE, non_blocking=True)
        probabilities = MODEL(images).softmax(dim=1)
        scores5, ids5 = probabilities.topk(5, dim=1)
        for j, local_id in enumerate(local_targets.tolist()):
            idx = EVAL_INDICES[current+j]
            true_id = IMAGENET_ID_BY_LOCAL[local_id]
            pred_id = int(ids5[j,0].item())
            top5_ids = ids5[j].tolist()
            rows.append({
                'dataset_index': idx,
                'filename': str(Path(VAL_DATASET.samples[idx][0]).relative_to(DATA_ROOT)),
                'true_imagenette_id': local_id,
                'true_imagenet1k_id': true_id,
                'true_class': CLASS_NAMES_10[local_id],
                'pred_imagenet1k_id': pred_id,
                'pred_class': IMAGENET_CLASSES[pred_id],
                'pred_confidence': float(scores5[j,0].item()),
                'top5_imagenet1k_ids': ' '.join(map(str, top5_ids)),
                'top1_correct': int(pred_id == true_id),
                'top5_correct': int(true_id in top5_ids),})
        current += len(local_targets)
PREDICTIONS = pd.DataFrame(rows)
TOP1 = PREDICTIONS['top1_correct'].mean()
TOP5 = PREDICTIONS['top5_correct'].mean()
print(f'sample N={len(PREDICTIONS)}，Top-1={TOP1:.4%}，Top-5={TOP5:.4%}，time {time.time()-started:.1f}s')
if QUICK_TEST:
    print('Now is QUICK_TEST')
else:
    print(' Finish 3925 picture evaluation')
PREDICTIONS.to_csv(OUTPUT_DIR/'resnet18_imagenette_val_predictions.csv', index=False)
display(PREDICTIONS.head(8))
````

## 3 Feature Map Visualization and Class Activation Mapping (CAM)

### 3.1 原理与解释对象

本实验通过可视化 ResNet18 的中间特征图，观察不同通道对输入图像的空间响应；进一步采用 CAM，将卷积特征与分类器权重结合，构造针对预测类别的空间重要性热图，该实现属于基于已有输入的激活可视化与类别归因

````python
class FeatureVisualizationCAM:
    def __init__(self, model, layer=None):
        self.model = model
        self.layer = layer if layer is not None else model.layer4
        if not (hasattr(model, 'fc') and hasattr(model.fc, 'weight')):
            raise ValueError('CAM requires a GAP followed by a linear classification layer; the current implementation is for ResNet18')

    @torch.no_grad()
    def extract(self, input_tensor, target_class):
        if input_tensor.ndim != 4 or input_tensor.shape[0] != 1:
            raise ValueError('This interface only processes one image at a time: [1,3,H,W]')
        activations = []
        handle = self.layer.register_forward_hook(lambda m, i, o: activations.append(o.detach()))
        try:
            logits = self.model(input_tensor)
        finally:
            handle.remove()
        feats = activations[0][0]  # [512,7,7]
        weights = self.model.fc.weight[int(target_class)]
        if feats.shape[0] != weights.numel():
            raise ValueError('The number of CAM feature channels does not match the classifier weights')
        signed_cam = torch.einsum('c,chw->hw', weights, feats)
        cam = signed_cam.clamp_min(0)
        energy = feats.square().mean(dim=0).sqrt()

        def resize_normalize(two_d):
            up = F.interpolate(two_d[None,None], size=input_tensor.shape[-2:],
                               mode='bilinear', align_corners=False)[0,0]
            up = up - up.min()
            return (up / up.max().clamp_min(1e-8)).cpu().numpy().astype(np.float32)

        return {'features': feats.cpu().numpy(),
                'energy': resize_normalize(energy),
                'cam': resize_normalize(cam),
                'prediction': int(logits.argmax(dim=1).item())}

FV = FeatureVisualizationCAM(MODEL)
MEAN = torch.tensor([0.485,0.456,0.406]).view(1,3,1,1)
STD  = torch.tensor([0.229,0.224,0.225]).view(1,3,1,1)

def show_rgb(input_tensor):
    """Denormalization for visualization, without modifying the input tensor."""
    a = input_tensor.detach().cpu().float()
    rgb = (a * STD + MEAN).clamp(0,1)
    return rgb[0].permute(1,2,0).numpy()

def fv_explainer(input_tensor, target_class):
    """[Unified Interface] XAI(input_tensor[1,3,H,W], int target) -> Non-negative [H,W] importance map."""
    return FV.extract(input_tensor, target_class)['cam']

# Pick the first correctly classified image from each different category
VIS_INDICES = []
for local_class in range(10):
    found = PREDICTIONS[(PREDICTIONS.true_imagenette_id==local_class) &
                        (PREDICTIONS.top1_correct==1)]
    if len(found): VIS_INDICES.append(int(found.iloc[0].dataset_index))
VIS_INDICES = VIS_INDICES[:6]
assert VIS_INDICES, 'Without correctly predicted samples, it is impossible to create a visual demonstration.'
print('Feature Visualization sample index：', VIS_INDICES)
````

````python
# 3.3 Different Categories: Original Image / Feature Intensity / CAM / Overlay
fig, axes = plt.subplots(len(VIS_INDICES), 4, figsize=(13, 3.1*len(VIS_INDICES)), squeeze=False)
for row_idx, idx in enumerate(VIS_INDICES):
    x, local_target = VAL_DATASET[idx]
    x = x.unsqueeze(0).to(DEVICE)
    target = int(PREDICTIONS.loc[PREDICTIONS.dataset_index==idx, 'pred_imagenet1k_id'].iloc[0])
    result = FV.extract(x, target)
    rgb = show_rgb(x)
    for col, image in enumerate([rgb, result['energy'], result['cam'], rgb]):
        axis = axes[row_idx,col]
        if col in [0,3]: axis.imshow(image)
        else: axis.imshow(image, cmap='magma', vmin=0, vmax=1)
        if col == 3: axis.imshow(result['cam'], cmap='jet', alpha=0.43, vmin=0, vmax=1)
        axis.axis('off')
    axes[row_idx,0].set_title(f'Original: {CLASS_NAMES_10[local_target]}')
    axes[row_idx,1].set_title('Feature energy (class-agnostic)')
    axes[row_idx,2].set_title(f'CAM for {IMAGENET_CLASSES[target]}')
    axes[row_idx,3].set_title('CAM overlay')
fig.tight_layout(); fig.savefig(OUTPUT_DIR/'feature_visualization_cam.png', dpi=160, bbox_inches='tight')
plt.show()
````

**结果分析:**

可视化结果表明，模型倾向于关注与目标类别密切相关的局部视觉结构。 对于 Tench，CAM的主要响应集中于中部及下部；对于Chain Saw，高响应区域主要位于机身中部和右侧发动机外壳；对于Church，模型主要关注左上方钟楼，以及部分屋顶邻近区域；对于French Horn，模型更关注中右部盘绕的金属管道,对于Cassette Player，模型关注正面的控制面板、磁带仓及部分扬声器区域，对于English Springer，模型关注犬的身体、腿部及部分相邻犬体。这说明 ResNet18的分类过程并不均匀关注整个物体，是利用具有区分性的局部特征形成预测。

````python
# 3.4 Single-channel feature map: Select the 8 channels in layer4 with the highest average activation
idx = VIS_INDICES[0]
x, _ = VAL_DATASET[idx]
x = x.unsqueeze(0).to(DEVICE)
target = int(PREDICTIONS.loc[PREDICTIONS.dataset_index==idx, 'pred_imagenet1k_id'].iloc[0])
detail = FV.extract(x, target)
features = detail['features']
channel_rank = np.argsort(features.mean(axis=(1,2)))[-8:][::-1]
fig, axes = plt.subplots(2, 5, figsize=(14,5.8))
axes[0,0].imshow(show_rgb(x)); axes[0,0].set_title('Original'); axes[0,0].axis('off')
axes[0,1].imshow(detail['cam'], cmap='magma'); axes[0,1].set_title('Class CAM'); axes[0,1].axis('off')
for j,ch in enumerate(channel_rank):
    ax = axes.flat[j+2]
    ax.imshow(features[ch], cmap='magma', interpolation='nearest')
    ax.set_title(f'layer4 channel #{ch}')
    ax.axis('off')
fig.suptitle('Top activated channels (native resolution 7x7)')
fig.tight_layout(); fig.savefig(OUTPUT_DIR/'top_feature_channels.png', dpi=160, bbox_inches='tight')
plt.show()
````

**单通道特征图分析:**

单通道Feature Visualization 实验图显示了Tench图片在 layer4中平均激活最强的八个特征通道。我们可以看到，不同通道的最高响应位置不同，有些集中于鱼身，有些集中于图像上方，也有些同时覆盖多个空间区域。

## 4忠诚度评估

目的：如果XAI判断区域很重要，验证修改该区域是否确实显著影响模型对原预测类别的置信度。
我主要采用：

- Deletion：从原图出发，按 XAI 重要性从高到低依次遮蔽区域，记录固定目标类别概率,下降越快 / AUC 越低越好。
- Insertion：从基线图出发，按重要性从高到低逐步恢复区域，记录同一类别概率,上升越快 / AUC 越高越好。
- Random baseline：随机打乱块的顺序，使用相同遮蔽方式与相同模型。

### 公平性约定（组员共享）

1. 固定模型、权重、224×224 预处理、val 图像及样本顺序。
2. 固定目标类为未扰动原图的 1000 类 Top-1，且全过程不重新选类别。
3. 统一以14×14个不重叠patch排序（每块 16×16 像素），按 `steps=14` 等量扰动，避免只比较不同分辨率热力图的插值噪声。
4. 基线是归一化张量中的0，对应输入 RGB 空间的 ImageNet 三通道均值，不是纯黑 RGB。
5. `softmax` 概率在 `[0,1]`，曲线横轴为扰动图像的面积比例，使用 `np.trapezoid` / `np.trapz` 计算 AUC；**所有 XAI 方法共用相同样本和随机基线**。
6. 解释器必须返回与输入空间一致的非负重要性 `[H,W]`；对于 IG 等带符号方法，可先用 `attribution.clamp_min(0)`（正证据）或 `abs()`（绝对贡献），但需要在报告中写明并对所有比较结果说明含义。

````python
# 4.1 通用遮蔽 / Deletion / Insertion 计算

def as_nonnegative_map(saliency, height, width, signed_mode='positive'):
    """兼容 numpy / torch，输入 [H,W] 或 [1,H,W] 或 [1,1,H,W]。
    signed_mode='positive': 只看正贡献；'absolute': 看绝对贡献。
    """
    if isinstance(saliency, torch.Tensor):
        a = saliency.detach().cpu().float().numpy()
    else:
        a = np.asarray(saliency, dtype=np.float32)
    a = np.squeeze(a)
    if a.ndim != 2: raise ValueError(f'XAI 需要返回二维空间图，收到 {a.shape}')
    if not np.isfinite(a).all(): raise ValueError('XAI 图出现 NaN / Inf')
    if signed_mode == 'positive': a = np.maximum(a, 0)
    elif signed_mode == 'absolute': a = np.abs(a)
    else: raise ValueError('signed_mode 只能是 positive / absolute')
    t = torch.from_numpy(a.copy())[None,None]
    if a.shape != (height,width):
        t = F.interpolate(t, size=(height,width), mode='bilinear', align_corners=False)
    return t[0,0].numpy()


def pixel_order_to_patch_order(saliency, shape=(224,224), patch_grid=14, signed_mode='positive'):
    """重要性图 → patch 重要性降序排列，越前的 patch 越重要。"""
    h,w = shape
    assert h % patch_grid == 0 and w % patch_grid == 0, '输入宽高必须能被 patch_grid 整除'
    importance = as_nonnegative_map(saliency, h, w, signed_mode)
    patch_values = F.adaptive_avg_pool2d(torch.from_numpy(importance)[None,None],
                                         (patch_grid,patch_grid)).flatten().numpy()
    return np.argsort(-patch_values, kind='stable')


@torch.no_grad()
def perturbation_curve(model, input_tensor, target_class, ordered_patches,mode='deletion', steps=14, patch_grid=14):
    """根据给定 patch 顺序输出面积比例 vs 固定目标类 softmax 概率。
    0%: deletion=原图；insertion=全基线。100%: 两者互换。
    """
    if mode not in ('deletion','insertion'): raise ValueError('mode: deletion or insertion')
    assert input_tensor.ndim==4 and input_tensor.shape[0]==1
    h,w=input_tensor.shape[-2:]
    assert h%patch_grid==0 and w%patch_grid==0
    total=patch_grid**2
    order=np.asarray(ordered_patches, dtype=np.int64)
    assert len(order)==total and set(order.tolist())==set(range(total)), 'The patch order must be a sequence from 0 to grid²-1'
    rank=np.empty(total, dtype=np.int64)
    rank[order]=np.arange(total)
    rank=torch.from_numpy(rank).to(input_tensor.device).reshape(patch_grid,patch_grid)
    rank=rank.repeat_interleave(h//patch_grid,0).repeat_interleave(w//patch_grid,1)
    fractions=np.linspace(0,1,steps+1)
    levels=np.linspace(0,total,steps+1).round().astype(int)
    masks=torch.stack([(rank < int(k)) for k in levels])[:,None].float()  # [S,1,H,W]
    if mode=='deletion':
        variants=input_tensor*(1-masks)
    else:
        variants=input_tensor*masks
    batch_size=12 if input_tensor.device.type=='cpu' else 32
    scores=[]
    for batch in variants.split(batch_size,dim=0):
        scores.extend(model(batch).softmax(1)[:,int(target_class)].cpu().numpy().tolist())
    return fractions, np.array(scores, dtype=float)

def auc_of_curve(frac, conf):
    return float(np.trapezoid(conf,frac)) if hasattr(np,'trapezoid') else float(np.trapz(conf,frac))
````

````python
# 4.2 General Evaluator: Using the same batch of images, the same target category, and the same random baseline for all methods
# explainers is dict: method name -> fn(input_tensor[1,3,224,224], target:int) -> [224,224]

def evaluate_xai_methods(explainers, dataset, predictions, sample_indices,
                         steps=14, patch_grid=14, random_repeats=3, seed=2026,
                         signed_mode='positive'):
    if not explainers: raise ValueError('at least one XAI method')
    if len(sample_indices)==0: raise ValueError('at least one picture')
    pred_lookup=predictions.set_index('dataset_index')
    all_curves={k:{'deletion':[], 'insertion':[]} for k in explainers}
    all_curves['Random']={'deletion':[], 'insertion':[]}
    records=[]
    fractions=np.linspace(0,1,steps+1)
    for n,idx in enumerate(sample_indices,1):
        x,_=dataset[idx]; x=x[None].to(DEVICE)
        target=int(pred_lookup.loc[idx,'pred_imagenet1k_id'])
        h,w=x.shape[-2:]
        # The random baseline is shared across all methods to ensure a fair comparison.
        rng=np.random.default_rng(seed+int(idx))
        random_order_list=[rng.permutation(patch_grid**2) for _ in range(random_repeats)]
        random_pair=[]
        for order in random_order_list:
            random_pair.append({mode:perturbation_curve(MODEL,x,target,order,mode,steps,patch_grid)[1]for mode in ('deletion','insertion')})
        for mode in ('deletion','insertion'):
            random_avg=np.mean([p[mode] for p in random_pair],axis=0)
            all_curves['Random'][mode].append(random_avg)
            records.append({'dataset_index':idx,'method':'Random','mode':mode,'auc':auc_of_curve(fractions,random_avg)})
        for name,explainer in explainers.items():
            importance=explainer(x,target)
            order=pixel_order_to_patch_order(importance,shape=(h,w),patch_grid=patch_grid,signed_mode=signed_mode)
            for mode in ('deletion','insertion'):
                _, curve=perturbation_curve(MODEL,x,target,order,mode,steps,patch_grid)
                all_curves[name][mode].append(curve)
                records.append({'dataset_index':idx,'method':name,'mode':mode,'auc':auc_of_curve(fractions,curve)})
        print(f'Fidelity evaluated {n}/{len(sample_indices)}: idx={idx}, target={IMAGENET_CLASSES[target]}')
    curve_means={name:{mode:np.mean(arrs,axis=0) for mode,arrs in values.items()}for name,values in all_curves.items()}
    auc_df=pd.DataFrame(records)
    summary=auc_df.groupby(['method','mode'])['auc'].agg(['mean','std','count']).reset_index()
    return {'fractions':fractions, 'mean_curves':curve_means, 'sample_curves':all_curves,'per_image_auc':auc_df, 'summary':summary, 'indices':list(sample_indices)}

def plot_fidelity_comparison(evaluation, save_path=None):
    fig,axes=plt.subplots(1,2,figsize=(13,4.8))
    for ax,mode in zip(axes,['deletion','insertion']):
        for name,curve in evaluation['mean_curves'].items():
            style='--' if name=='Random' else '-'
            mean_auc=auc_of_curve(evaluation['fractions'],curve[mode])
            ax.plot(evaluation['fractions'],curve[mode],style,linewidth=2,
                    label=f'{name} (AUC={mean_auc:.3f})')
        ax.set(xlabel='Fraction of patches perturbed',ylabel='P(original predicted class)',
               xlim=(0,1),ylim=(0,1),title=f'{mode.title()} (lower AUC is better)' if mode=='deletion'
               else 'Insertion (higher AUC is better)')
        ax.grid(alpha=.2); ax.legend(fontsize=9)
    fig.suptitle(f'Faithfulness | N={len(evaluation["indices"])} images | same 1000-class model')
    fig.tight_layout()
    if save_path: fig.savefig(save_path, dpi=160, bbox_inches='tight')
    plt.show()
````

## 5 同一个 CNN 在不同条件下，关注区域是否稳定？

该演示不改变空间位置的亮度与模糊变换，使原图和变换图的 CAM 能在相同像素坐标中比较。这里用空间Pearson相关系数衡量热图一致性，范围为 `[-1,1]`，越接近 1 说明两张热图形状越相似。我们使用原图预测类别作为所有变换条件的固定解释目标；另列变换后预测是否发生变化。同时，所有变换都从同一张 224×224、经过中心裁剪的 RGB 原图出发，使用相同 ImageNet 归一化。

````python
# 5.1 Brightness / Blur without geometric distortion: small sample check
def change_conditions(x):
    rgb=torch.from_numpy(show_rgb(x)).permute(2,0,1)  # [3,H,W] in [0,1]
    mean=MEAN[0]; std=STD[0]
    original=x
    dark=(TF.adjust_brightness(rgb,0.70)[None]-mean)/std
    bright=(TF.adjust_brightness(rgb,1.30)[None]-mean)/std
    blurry=(TF.gaussian_blur(rgb,kernel_size=[9,9],sigma=[1.8,1.8])[None]-mean)/std
    return {'original': original,
            'brightness 0.7':dark.to(x.device),
            'brightness 1.3':bright.to(x.device),
            'gaussian blur':blurry.to(x.device)}
def spatial_pearson(a,b):
    a=np.asarray(a).ravel().astype(float); b=np.asarray(b).ravel().astype(float)
    if a.std()<1e-12 or b.std()<1e-12: return np.nan
    return float(np.corrcoef(a,b)[0,1])

condition_records=[]
idx=FIDELITY_INDICES[0]
x,_=VAL_DATASET[idx]; x=x[None].to(DEVICE)
original_target=int(PREDICTIONS.set_index('dataset_index').loc[idx,'pred_imagenet1k_id'])
variations=change_conditions(x)
original_map=fv_explainer(variations['original'],original_target)
fig, axes = plt.subplots(2,len(variations),figsize=(13,6))
for j,(cond,xcond) in enumerate(variations.items()):
    explain=FV.extract(xcond,original_target)
    new_pred=explain['prediction']
    correlation=spatial_pearson(original_map,explain['cam'])
    condition_records.append({'condition':cond,'cam_correlation':correlation,'pred_after':IMAGENET_CLASSES[new_pred],'pred_unchanged':new_pred==original_target})
    axes[0,j].imshow(show_rgb(xcond)); axes[0,j].axis('off'); axes[0,j].set_title(cond)
    axes[1,j].imshow(show_rgb(xcond)); axes[1,j].imshow(explain['cam'],cmap='jet',alpha=.43,vmin=0,vmax=1)
    axes[1,j].axis('off'); axes[1,j].set_title(f'CAM corr={correlation:.3f}\nlabel kept={new_pred==original_target}')
fig.suptitle('Feature Visualization under appearance changes (same fixed target)')
fig.tight_layout(); fig.savefig(OUTPUT_DIR/'feature_stability_conditions.png',dpi=160,bbox_inches='tight')
plt.show()
display(pd.DataFrame(condition_records))
````

**结果分析：**

在忠实性评估方面，本实验采用 Deletion和Insertion两种扰动方法，并将CAM与随机区域排序进行比较。实验共选择10张正确分类的图片，每个Imagenette类别一张。结果显示，CAM的DeletionAUC为0.1143，低于随机基线的0.1426；InsertionAUC为0.5725，明显高于随机基线的0.1253。
Deletion曲线显示，优先移除CAM 高亮区域能够更快降低目标类别的预测置信度,Insertion曲线则表明，优先恢复CAM高亮区域能够更快恢复模型的分类置信度。尤其是在仅恢复约36%图像区域时，CAM的平均预测置信度已达到约0.64，而随机恢复的置信度仍接近零。这表明CAM所识别的高重要性区域确实包含对模型分类具有较强影响的视觉信息。
此外，亮度扰动实验表明，CAM在所测试的Tench图片上保持了较高的空间相关性（0.9978和0.9990）。然而，在高斯模糊条件下，尽管CAM 相关系数仍达到0.9446，模型的预测类别却发生了变化。这说明关注区域的稳定性并不必然意味着预测结果稳定。
总体而言，Feature Visualization（CAM）能够提供具有一定可解释性且与模型预测行为相符的空间解释。在本次实验中，该方法的Deletion和Insertion结果均优于随机基线，尤其是在Insertion指标上表现突出。

**Feature Visualization的局限性**

Feature Visualization在这个研究中也有一定的局限性。首先，热力图空间分辨率较低,最后卷积层只有7×7的空间分辨率，之后通过插值放大到224×224，因此高亮区域的边界较模糊，不能视为精确的物体分割结果。其次，忠实性评估样本量有限,准确率是在全部3,925张验证图片上计算的，但忠实性仅在10张正确分类图片上评估，Insertion AUC 的样本标准差为 0.2290，说明不同图片之间存在明显差异，不能据此推断总体统计显著性。
